```python
import shutil
for dir_name, _ , files in os.walk(path):
    for filename in filenames:
        print(os.path.join(dir_name,filename))
        shutil.copy(os.path.join(dir_name,filename), data_dir_name)
```

```python
# src/data_ingestion.py
import pandas as pd

def load_and_combine_data(fake_file_path, true_file_path):
    # Load the fake and true news CSV files
    fake_df = pd.read_csv(fake_file_path)
    true_df = pd.read_csv(true_file_path)

    # Assign labels: 0 for fake news, 1 for real news
    fake_df['label'] = 0
    true_df['label'] = 1

    # Combine the two datasets into one DataFrame
    combined_df = pd.concat([fake_df, true_df], ignore_index=True)
    return combined_df

if __name__ == "__main__":
    fake_file = "../data/Fake.csv"
    true_file = "../data/True.csv"
    combined_data = load_and_combine_data(fake_file, true_file)
    combined_data.to_csv("../data/combined_news.csv", index=False)
    print("Combined data saved to data/combined_news.csv")
```
