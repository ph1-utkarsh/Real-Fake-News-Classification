```python
import re
import numpy as np 
import pandas as pd 
import nltk
from nltk.corpus import stopwords
from nltk.stem import WordNetLemmatizer
import os
from nltk.tokenize import word_tokenize
```

```python
nltk_data_path = r"C:\Users\hp\AppData\Roaming\nltk_data"  # Raw string is important for Windows paths
if os.path.exists(nltk_data_path):
    nltk.data.path.append(nltk_data_path)
    print("NLTK data path appended.")
else: # Download only if not present already.
    nltk.download('punkt')
    nltk.download('stopwords')
    nltk.download('wordnet')
    nltk.data.path.append(nltk_data_path)


def clean_text(text):
    text = text.lower()
    text = re.sub(r'[^a-zA-Z\s]', '', text)
    tokens = word_tokenize(text)
    stop_words = set(stopwords.words('english'))
    tokens = [word for word in tokens if word not in stop_words]
    lemmatizer = WordNetLemmatizer()
    tokens = [lemmatizer.lemmatize(word) for word in tokens]
    return " ".join(tokens)
```

```python
def preprocess_dataframe(df):
    data_to_clean = df.copy()
    # Apply the clean_text function to the 'text' column
    data_to_clean['clean_text'] = data_to_clean['text'].astype(str).apply(clean_text)
    return data_to_clean
```

```python
if __name__ == "__main__":
    try:
      df = pd.read_csv('../data/combined_news.csv')
      df_preprocessed = preprocess_dataframe(df)
      df_preprocessed.to_csv('../data/combined_news_preprocessed.csv', index=False)
      print(df_preprocessed.head())
      print("Shape after preprocessing", df_preprocessed.shape)
      print("Preprocessed data saved to data/combined_news_preprocessed.csv")
    except FileNotFoundError:
      print("Error: combined_news.csv not found in the data directory. Please make sure the file exists.")
    except Exception as e:
      print(f"An error occurred: {e}")
```
