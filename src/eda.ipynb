```python
import matplotlib.pyplot as plt
import seaborn as sns

def perform_eda(data_path):
    df = pd.read_csv(data_path)
    
    # Basic statistics
    print("Data Shape:", df.shape)
    print("Columns:", df.columns)
    print("Missing Values:\n", df.isnull().sum())
    
    # Class distribution
    plt.figure(figsize=(6, 4))
    sns.countplot(x='label', data=df)
    plt.title("Class Distribution (0: Fake, 1: Real)")
    plt.savefig("../data/class_distribution.png")
    plt.show()
    
    # Distribution of article lengths (based on the 'text' column)
    df['text_length'] = df['text'].apply(lambda x: len(str(x).split()))
    plt.figure(figsize=(8, 6))
    sns.histplot(df['text_length'], bins=50, kde=True)
    plt.title("Distribution of Article Lengths")
    plt.xlabel("Number of Words")
    plt.savefig("../data/article_length_distribution.png")
    plt.show()

if __name__ == "__main__":
   data_file = "../data/combined_news.csv"
   perform_eda(data_file)
```
